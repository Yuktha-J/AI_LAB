```python
def calculate_cost(board):
    conflicts = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):
            if (board[i] == board[j] or
                abs(board[i] - board[j]) == abs(i - j)):
                conflicts += 1

    return conflicts

def print_board(board):
    n = len(board)

    for row in range(n):
        for col in range(n):
            if board[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()

def hill_climbing(board):
    current_cost = calculate_cost(board)
    step = 0

    while current_cost != 0:
        print("\nStep:", step)
        print("Current state:", board)
        print("Current cost:", current_cost)
        best_board = board.copy()
        best_cost = current_cost

        for col in range(len(board)):
            for row in range(len(board)):
                if row != board[col]:
                    neighbor = board.copy()
                    neighbor[col] = row
                    cost = calculate_cost(neighbor)
                    print("Neighbor:", neighbor,
                          "Cost:", cost)
                    if cost < best_cost:
                        best_board = neighbor
                        best_cost = cost

        if best_cost == current_cost:
            print("Local minimum reached.")
            print("Try another initial state.")
            return

        board = best_board
        current_cost = best_cost
        print("Selected state:", board)
        print("Selected cost:", current_cost)

        step += 1

    print("\nGoal state reached!")
    print("Final state:", board)
    print("Final cost:", current_cost)
    print_board(board)

initial_board = [0, 0, 0, 0]
hill_climbing(initial_board)
```
