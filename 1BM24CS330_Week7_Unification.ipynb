```python
def is_variable(x):
    return x.islower()

def apply_substitution(x, substitution):
    while x in substitution:
        x = substitution[x]
    return x

def occurs(variable, expression, substitution):
    expression = apply_substitution(expression, substitution)
    if expression == variable:
        return True
    if "(" in expression:
        inside = expression[expression.index("(") + 1:-1]
        parts = inside.split(",")

        for part in parts:
            if occurs(variable, part, substitution):
                return True
    return False

def unify(x, y, substitution):
    x = apply_substitution(x, substitution)
    y = apply_substitution(y, substitution)
    if x == y:
        return substitution
    if is_variable(x):
        if occurs(x, y, substitution):
            return None
        substitution[x] = y
        return substitution
    if is_variable(y):
        if occurs(y, x, substitution):
            return None
        substitution[y] = x
        return substitution
    if "(" not in x or "(" not in y:
        return None
    predicate_x = x[:x.index("(")]
    predicate_y = y[:y.index("(")]
    if predicate_x != predicate_y:
        return None
    args_x = x[x.index("(") + 1:-1].split(",")
    args_y = y[y.index("(") + 1:-1].split(",")
    if len(args_x) != len(args_y):
        return None
    for a, b in zip(args_x, args_y):
        substitution = unify(
            a.strip(),
            b.strip(),
            substitution
        )
        if substitution is None:
            return None
    return substitution

print("UNIFICATION ALGORITHM")
print("---------------------")
x = input("Enter first expression: ")
y = input("Enter second expression: ")
substitution = {}
result = unify(x, y, substitution)
print()
if result is None:
    print("FAILURE")
elif len(result) == 0:
    print("NIL")
else:
    print("SUBSTITUTION:")
    for variable, value in result.items():
        value = apply_substitution(value, result)
        print(variable, "->", value)
```
