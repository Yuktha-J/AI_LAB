```python
def pl_true(exp, model):
    if isinstance(exp, str):
        return model[exp]

    op = exp[0]

    if op == "NOT":
        return not pl_true(exp[1], model)
    if op == "AND":
        return all(pl_true(x, model) for x in exp[1:])
    if op == "OR":
        return any(pl_true(x, model) for x in exp[1:])
    if op == "IMPLIES":
        return not pl_true(exp[1], model) or pl_true(exp[2], model)
    if op == "IFF":
        return pl_true(exp[1], model) == pl_true(exp[2], model)


def tt_check_all(KB, alpha, symbols, model):
    if not symbols:
        return not pl_true(KB, model) or pl_true(alpha, model)

    P = symbols[0]
    rest = symbols[1:]

    m1 = model.copy()
    m1[P] = True

    m2 = model.copy()
    m2[P] = False

    return (tt_check_all(KB, alpha, rest, m1)
            and tt_check_all(KB, alpha, rest, m2))


def tt_entails(KB, alpha):
    symbols = sorted(set(get_symbols(KB) | get_symbols(alpha)))
    return tt_check_all(KB, alpha, symbols, {})


def get_symbols(exp):
    if isinstance(exp, str):
        return {exp}
    return set().union(*(get_symbols(x) for x in exp[1:]))

KB = ("AND", ("IMPLIES", "P", "Q"), "P")
alpha = "Q"

if tt_entails(KB, alpha):
    print("KB entails Query")
else:
    print("KB does not entail Query")
```
